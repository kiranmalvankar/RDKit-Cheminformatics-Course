# Molecular Descriptor

## Why Need Molecular Descriptors?

Computers cannot understand molecules the way chemists do.

A chemist can look at a molecular structure and recognize:

- It is small.
- It is hydrophobic.
- It contains aromatic rings.
- It has hydrogen bond donors.

A computer cannot interpret these visual chemical features directly.

To analyze molecules computationally, chemical structures must first be converted into numerical values.

These numerical values are called molecular descriptors.

## What is a Molecular Descriptor?

A molecular descriptor is a numerical value that describes a specific chemical or physical property of a molecule.

Descriptors convert molecular structures into numbers that can be analyzed by computers.

They provide a quantitative representation of a molecule's characteristics and are widely used in cheminformatics, QSAR modeling, drug discovery, and machine learning.

# Type of Molecular Descriptor 

## Types of Molecular Descriptors

Molecular descriptors can be broadly classified into several categories based on the type of molecular information they represent.

### 1. Physicochemical Descriptors (Physical and Chemical Properties)

Describe measurable physicochemical properties of a molecule.

Examples:
- Molecular Weight
- LogP
- TPSA
- Molar Refractivity

---

### 2. Topological Descriptors (2D Molecular Connectivity)

Describe how atoms are connected within a molecule without considering 3D coordinates.

Examples:
- Rotatable Bonds
- Ring Count
- Chi Indices
- Kier Shape Indices

---

### 3. Electronic Descriptors (Electron Distribution and Charge Properties)

Describe the electronic characteristics of a molecule that influence chemical reactivity.

Examples:
- Partial Charges
- Dipole Moment
- HOMO/LUMO Energies
- Electronegativity-related Properties

---

### 4. Geometrical Descriptors (3D Shape and Spatial Properties)

Describe the three-dimensional geometry of a molecule.

Examples:
- Molecular Volume
- Surface Area
- Radius of Gyration
- Shape Descriptors

---

### 5. Constitutional Descriptors (Basic Structural Composition)

Describe simple structural information obtained directly from the molecular formula and atom counts.

Examples:
- Number of Atoms
- Heavy Atom Count
- Number of Bonds
- Number of Heteroatoms

## Essential RDKit Descriptors

In this notebook, we will focus on the most commonly used molecular descriptors in cheminformatics and drug discovery.

- Molecular Weight (Molecular Size)
- LogP (Lipophilicity / Hydrophobicity)
- TPSA (Molecular Polarity)
- Hydrogen Bond Acceptors (Hydrogen Bonding Ability)
- Hydrogen Bond Donors (Hydrogen Bonding Ability)
- Rotatable Bonds (Molecular Flexibility)
- Heavy Atom Count (Non-Hydrogen Atom Count)
- Ring Count (Molecular Cyclicity)

```python
# Example

from rdkit import Chem
from rdkit.Chem import Descriptors
from rdkit.Chem import Crippen
from rdkit.Chem import Lipinski
import pandas as pd

# List of compounds
compounds = {
    "Ethanol": "CCO",
    "Methanol": "CO",
    "Acetone": "CC(=O)C",
    "Acetic Acid": "CC(=O)O",
    "Benzene": "c1ccccc1",
    "Toluene": "Cc1ccccc1",
    "Phenol": "Oc1ccccc1",
    
}

# Empty list to store descriptor values
descriptor_data = []

# Calculate descriptors
for name, smiles in compounds.items():

    mol = Chem.MolFromSmiles(smiles)

    descriptor_data.append({
        "Compound": name,
        "MolWt": round(Descriptors.MolWt(mol), 2),
        "LogP": round(Crippen.MolLogP(mol), 2),
        "TPSA": round(Descriptors.TPSA(mol), 2),
        "HBA": Lipinski.NumHAcceptors(mol),
        "HBD": Lipinski.NumHDonors(mol),
        "RotBonds": Lipinski.NumRotatableBonds(mol),
        "HeavyAtoms": Lipinski.HeavyAtomCount(mol),
        "Rings": Lipinski.RingCount(mol)
    })

# Create DataFrame
df = pd.DataFrame(descriptor_data)

# Display the results
df
```

In a real cheminformatics workflow, descriptors are rarely calculated one at a time.

Instead, a set of molecular descriptors is generated for every molecule in a dataset, producing a descriptor matrix.

This descriptor matrix serves as the input for:

- QSAR Modeling
- Machine Learning
- ADMET Prediction
- Virtual Screening
- Drug-likeness Assessment
- Statistical Data Analysis

This workflow forms the foundation of modern AI-driven drug discovery.
