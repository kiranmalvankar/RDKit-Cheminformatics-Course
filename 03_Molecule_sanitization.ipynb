# Molecule Sanitization

# Why is sanitization necessary?
RDKit assumes that every molecule it processes should obey basic chemical rules. However, molecules imported from external databases or generated by software may contain errors such as invalid valence, incorrect bond types, or aromaticity issues. Sanitization acts as a quality control step, ensuring that only chemically valid molecules proceed to downstream analyses like QSAR, docking, or descriptor calculation.

```python
# import Libraries
from rdkit import Chem
```

```python
# creating a Molecule
mol= Chem.MolFromSmiles('CCO')
# MolFromSmiles() automatically Sanitizes molecules by default
```

```python
print(mol)
```

Since "CCO" is chemically valid, RDKit automatically sanitizes it during MolFromSmiles().

```python
# perform Explicit Sanitization
Chem.SanitizeMol(mol)
print('Molecule Sanitized Successfully')
```

```python
# Verify the Molecule
print('Number of Atoms:',mol.GetNumAtoms())
print('Number of Bonds:',mol.GetNumBonds())
```

```python
# example: Chemically invalid Molecule
smiles='C(C)(C)(C)(C)C'
mol=Chem.MolFromSmiles(smiles)
```

# Handle Invalid Inputs safely

```python
smiles_list = [
    "CCO",          # Valid
    "CC(O",         # Invalid syntax
    "C(C)(C)(C)(C)C"  # Invalid valence
]

for smiles in smiles_list:
    mol = Chem.MolFromSmiles(smiles)

    if mol is None:
        print(f"{smiles} --> Invalid molecule")
    else:
        print(f"{smiles} --> Valid molecule")
```
