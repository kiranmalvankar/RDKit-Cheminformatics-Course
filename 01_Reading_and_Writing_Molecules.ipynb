## Why Molecular Representations Matter?

Computers cannot directly understand chemical structures.

Therefore, molecules must be represented in machine-readable formats.

Common molecular formats:

1. SMILES
2. Canonical SMILES
3. MOL files
4. SDF files

RDKit converts these representations into Molecule Objects for further analysis.

## What is SMILES?

SMILES (Simplified Molecular Input Line Entry System) is a text representation of chemical structures.

Examples:

Methane : C

Ethane : CC

Ethanol : CCO

Benzene : c1ccccc1

Import Libraries

```python
from rdkit import Chem
from rdkit.Chem import Draw
```

## Create Molecules from SMILES

```python
methane = Chem.MolFromSmiles("C")
ethane = Chem.MolFromSmiles("CC")
ethanol = Chem.MolFromSmiles("CCO")
benzene = Chem.MolFromSmiles("c1ccccc1")
```

# Visualize Molecules

```python
Draw.MolsToGridImage(
    [methane, ethane, ethanol, benzene],
    legends=["Methane","Ethane","Ethanol","Benzene"]
)
```

## Molecule → SMILES Conversion

```python
mol = Chem.MolFromSmiles("CCO")

Chem.MolToSmiles(mol)
```

## Canonical SMILES

A molecule can have multiple valid SMILES.

Canonical SMILES provides a unique standardized representation.

## Example_Ethanol:RDKit converts both structures to the same canonical representation.

mol1 = Chem.MolFromSmiles("CCO")
mol2 = Chem.MolFromSmiles("OCC")

print(Chem.MolToSmiles(mol1))
print(Chem.MolToSmiles(mol2))

# Mol Files
# What is a MOL File?
A MOL file is a chemical file format used to store the structure of a single molecule.
It contains:
The atoms present in the molecule.
The bonds connecting the atoms.
2D or 3D atomic coordinates.
Molecular connectivity.
Unlike SMILES, a MOL file stores the exact arrangement of atoms in space.

## Reading MOL Files

mol = Chem.MolFromMolFile("sample.mol")

## Writing MOL Files
mol = Chem.MolFromSmiles("CCO")

Chem.MolToMolFile(mol, "ethanol.mol")

# SDF files

What is an SDF File?
An SDF (Structure Data File) is a standard chemical file format used to store one or more molecules along with their associated properties. Unlike a SMILES string, an SDF file can contain: Molecular structure (2D or 3D coordinates)

Compound name

Molecular properties

Experimental data

Activity values

IDs

Any custom annotations

Think of it as a spreadsheet for molecules, where each row is a molecule and each molecule carries both its structure and metadata.
