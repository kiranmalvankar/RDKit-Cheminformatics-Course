## Why Molecule Objects Matter

Every operation in RDKit begins with a Molecule (Mol) object.

Whether you are:

- Calculating molecular descriptors
- Generating fingerprints
- Performing similarity searches
- Running QSAR models
- Conducting scaffold analysis
- Generating conformers

You must first create a Mol object.

Understanding the Mol object is therefore essential for mastering RDKit.

A Molecule Object (Mol) is RDKit's internal representation of a chemical structure.

Instead of storing a molecule as plain text (SMILES), RDKit converts it into a structured object containing:

- Atoms
- Bonds
- Ring information
- Molecular properties
- Stereochemistry
- Conformer information

# Why does RDKit use a Mol object instead of directly manipulating SMILES strings?
A SMILES string is simply a text representation of a molecule. While it is compact and easy to store, it cannot directly answer chemical questions such as:
How many atoms are present?
Which atoms are connected?
Is the molecule aromatic?
How many rings does it contain?
RDKit converts the SMILES into a Mol object, which acts like a digital model of the molecule. Once this model exists, RDKit can efficiently calculate descriptors, generate fingerprints, search substructures, and perform many other cheminformatics operations.

```python
from rdkit import Chem
mol= Chem.MolFromSmiles('CCO')
```

```python
type(mol)
```

## What information does a Mol object contain?
It stores:
Number of atoms,
Number of bonds,
Ring information,
Atom properties,
Bond properties,
Coordinates (if available),
Molecular properties.

```python
mol.GetNumAtoms()
```

```python
mol.GetNumBonds()
```

# Atom Object
Each atom inside a molecule is itself an object.

# Atom Properties

```python
for atom in mol.GetAtoms():
    print(atom)
```

```python
for atom in mol.GetAtoms():
    print(atom.GetIdx(),atom.GetSymbol())
```

# Bond Objects
Evry bond is also an object

```python
for bond in mol.GetBonds():
    print(
        bond.GetBeginAtomIdx(),
        bond.GetEndAtomIdx(),
        bond.GetBondType()
    )
```
