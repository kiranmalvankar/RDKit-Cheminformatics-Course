# Molecular Similarity

## What is Molecular Similarity?

Molecular similarity is a quantitative measure of how closely two molecules resemble each other based on their structural features.

Similarity is typically calculated by comparing molecular fingerprints using a similarity coefficient.

The most commonly used similarity measure in cheminformatics is the Tanimoto Similarity.

## Why Do We Need Molecular Similarity?

Drug discovery often begins with a known active molecule.

Medicinal chemists then search for molecules that are structurally similar, hoping they may exhibit similar biological activity.

Instead of comparing molecular structures manually, computers compare molecular fingerprints.

This process is known as molecular similarity searching.

✓ Molecular similarity compares chemical structures computationally.

✓ Fingerprints are used as the structural representation.

✓ Similarity is expressed as a numerical score.

✓ The most widely used similarity metric is the Tanimoto coefficient.

## What is Tanimoto Similarity?

Tanimoto Similarity is a numerical measure of structural similarity between two molecules based on their molecular fingerprints.

It compares the structural features present in both molecules and produces a similarity score ranging from:

0 → Completely different

1 → Identical fingerprints

Higher values indicate greater structural similarity.

## Tanimoto Similarity Formula

The Tanimoto coefficient is calculated as:

Similarity = (Common Features) / (Total Unique Features)

where:

- Common Features = Structural features present in both molecules.
- Total Unique Features = All structural features present in either molecule.

The score always ranges between 0 and 1.

# Practical Example
In this example, we will compare three different pairs of molecules using Morgan fingerprints and Tanimoto Similarity.

Our goal is to understand how structural similarity influences the similarity score.

We will compare:

1. Two very similar molecules
2. Two moderately similar molecules
3. Two structurally different molecules

```python
# Import Libraries
from rdkit import Chem
from rdkit.Chem import rdFingerprintGenerator
from rdkit import DataStructs
```

```python
# Create Morgan Fingerprints Generator
morgan_gen=rdFingerprintGenerator.GetMorganGenerator(radius=2,fpSize=2048)
```

```python
#very Similar Molecules
from rdkit import Chem
from rdkit.Chem import rdFingerprintGenerator
from rdkit import DataStructs

Paracetamol = Chem.MolFromSmiles("CC(=O)NC1=CC=C(C=C1)O")
Patacetamol = Chem.MolFromSmiles("CC(=O)NC1=CC=C(C=C1)O")

morgan_gen = rdFingerprintGenerator.GetMorganGenerator(
    radius=2,
    fpSize=2048
)

fp1 = morgan_gen.GetFingerprint(Paracetamol)
fp2 = morgan_gen.GetFingerprint(Paracetamol)

similarity = DataStructs.TanimotoSimilarity(fp1, fp2)

print(f"Tanimoto Similarity: {similarity:.3f}")
```

```python
paracetamol = Chem.MolFromSmiles("CC(=O)NC1=CC=C(C=C1)O")
aspirin = Chem.MolFromSmiles("CC(=O)Oc1ccccc1C(=O)O")

fp1 = morgan_gen.GetFingerprint(paracetamol)
fp2 = morgan_gen.GetFingerprint(aspirin)

similarity = DataStructs.TanimotoSimilarity(fp1, fp2)

print(f"Tanimoto Similarity: {similarity:.3f}")
```

```python
glucose = Chem.MolFromSmiles("OC[C@H]1O[C@@H](O)[C@H](O)[C@@H](O)[C@H]1O")
naphthalene = Chem.MolFromSmiles("c1cccc2ccccc12")

fp1 = morgan_gen.GetFingerprint(glucose)
fp2 = morgan_gen.GetFingerprint(naphthalene)

similarity = DataStructs.TanimotoSimilarity(fp1, fp2)

print(f"Tanimoto Similarity: {similarity:.3f}")
```
