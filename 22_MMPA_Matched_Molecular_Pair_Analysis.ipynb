# What is Matched Molecular Pair Analysis (MMPA)?

A Matched Molecular Pair (MMP) consists of two molecules that differ by only a small, well-defined structural change while sharing the same molecular core.

The structural change is called a transformation.

By comparing the properties of matched molecular pairs, medicinal chemists can understand how specific structural modifications influence biological activity and physicochemical properties.

# Workflor

```python
# EXample
# Import Libraries

from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem import Descriptors
import pandas as pd
```

```python
# Create Two Molecules

toluene = Chem.MolFromSmiles("Cc1ccccc1")

chlorobenzene = Chem.MolFromSmiles("Clc1ccccc1")
```

```python
# Visualize Molecules

Draw.MolsToGridImage(
    [toluene, chlorobenzene],
    legends=["Toluene", "Chlorobenzene"],
    molsPerRow=2,
    subImgSize=(300,300)
)
```

```python
# Define the Transformation

print("Transformation:")
print("CH3  →  Cl")
```

```python
# Compare Molecular Properties

comparison = pd.DataFrame({
    "Property":[
        "Molecular Weight",
        "LogP",
        "TPSA",
        "H-Bond Donors",
        "H-Bond Acceptors"
    ],
    "Toluene":[
        Descriptors.MolWt(toluene),
        Descriptors.MolLogP(toluene),
        Descriptors.TPSA(toluene),
        Descriptors.NumHDonors(toluene),
        Descriptors.NumHAcceptors(toluene)
    ],
    "Chlorobenzene":[
        Descriptors.MolWt(chlorobenzene),
        Descriptors.MolLogP(chlorobenzene),
        Descriptors.TPSA(chlorobenzene),
        Descriptors.NumHDonors(chlorobenzene),
        Descriptors.NumHAcceptors(chlorobenzene)
    ]
})

comparison
```

## Interpretation

The two molecules differ by only one structural modification:

CH3 → Cl

This is called a matched molecular pair transformation.

After replacing the methyl group with chlorine:

- Molecular weight increases.
- LogP changes.
- The aromatic core remains unchanged.

By comparing molecular properties before and after a single structural modification, medicinal chemists can understand the effect of specific transformations on molecular behavior.

This principle forms the foundation of Matched Molecular Pair Analysis (MMPA).
