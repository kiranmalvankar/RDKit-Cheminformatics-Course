# What is an SDF File?


An SDF (Structure Data File) is a standard chemical file format used to store one or more molecules along with their associated properties.
Unlike a SMILES string, an SDF file can contain:
Molecular structure (2D or 3D coordinates)

Compound name

Molecular properties

Experimental data

Activity values

IDs

Any custom annotations

Think of it as a spreadsheet for molecules, where each row is a molecule and each molecule carries both its structure and metadata.

Why do we need SDF files?

Suppose you download molecules from:

PubChem

ChEMBL

DrugBank

Zinc

Internal pharmaceutical databases
You will usually receive an SDF file, not a CSV of SMILES.

# Example

```python
# Import Libraries

from rdkit import Chem
from rdkit.Chem import Draw
```

```python
# Read the SDF file

supplier = Chem.SDMolSupplier("sdf_dataset.sdf")
```

```python
# Read the Molecules

molecules = [mol for mol in supplier if mol is not None]

print(f"Number of molecules: {len(molecules)}")
```

```python
# Display the Molecules

Draw.MolsToGridImage(
    molecules,
    molsPerRow=2,
    subImgSize=(300,300)
)
```

```python
# Print SMILES of Each Molecule

for i, mol in enumerate(molecules, start=1):
    smiles = Chem.MolToSmiles(mol)
    print(f"Molecule {i}: {smiles}")
```

```python
# Print all available properties

for i, mol in enumerate(molecules, start=1):

    print(f"\n----- Molecule {i} -----")

    for prop in mol.GetPropNames():
        print(prop, ":", mol.GetProp(prop))
```

## Interpretation

The SDF file was successfully loaded into RDKit. The molecular structure was extracted, visualized, converted to a SMILES representation, and any associated molecular properties were accessed. Finally, the molecule was written to a new SDF file, demonstrating how RDKit can both read and write SDF datasets.

SDF files are the standard format for exchanging molecular structures and associated metadata in cheminformatics, molecular modeling, QSAR studies, and drug discovery workflows.
