# What is Shape Similarity? 
Shape similarity measures how closely the three-dimensional shapes (volumes) of two molecules overlap after they are aligned.
Unlike fingerprints, which compare the presence or absence of chemical features, shape similarity compares the actual 3D volume occupied by the molecules.

 # Why is it important?
Two molecules can have:
Different chemical structures
Different substituents
Different fingerprints
but still have very similar 3D shapes.
If their shapes are similar, they may bind to the same protein.

# Applications

Shape-based virtual screening

Scaffold hopping

Lead optimization

Bioisostere identification

Drug repurposing

# Practical workflow

Generate 3D conformers.

Align the molecules.

Calculate the Shape Tanimoto Similarity.

Calculate the Shape Protrude Distance.

Visualize the aligned molecules.

Interpret the scores.

# Two Important metrics

# 1. Shape Tanimoto Similarity
0 → Completely different

1 → Identical shape

Higher values indicate greater shape overlap

# 2. Shape Protrude Distance

Measures how much one molecule sticks out beyond the other.

0 → Perfect overlap

Larger value → More protrusion

Lower values are better.
